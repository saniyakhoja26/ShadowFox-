## Research Question

Which product category generates the highest total sales and profit?


```python
category_summary = data.groupby("Category")[["Sales", "Profit"]].sum()
```

```python
category_summary
```

```python
category_summary = category_summary.reset_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x="Category", y="Sales", data=category_summary)
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x="Category", y="Profit", data=category_summary)
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.show()
```

## Conclusions and Insights

From the analysis of the sales dataset, the following insights were observed:

- The **Electronics** category generates the highest total sales among all categories.
- The **Electronics** category also contributes the highest overall profit.
- **Furniture** shows moderate sales and profit performance.
- **Office Supplies** has comparatively lower sales and profit but maintains consistent performance.

### Final Insight:
Electronics is the strongest performing category in terms of both sales and profit. 
Focusing more on Electronics-related products could help improve overall business revenue and profitability.
